```python
import pandas as pd

df = {
    "Studs":["Arjun","Kiran","Megha","Sanjay","Divya"],
    "Marks":[88,74,93,81,85],
    "Age":[19,22,21,23,20]
}

df = pd.DataFrame(df)

df
```

```python
plt.figure(figsize=(7,4),facecolor="blue") # it creates a canvas on which will show tha graph 
plt.bar(df["Studs"],df["Marks"])
plt.title("Data Viz")
plt.show()
```

```python
import matplotlib.pyplot as plt 
import pandas as pd 
```

```python
months=["jan","feb","mar","apr","may"]
sales=[1500,2000,1600,3000,4500]

plt.figure(figsize=(7,4),facecolor="white")

plt.plot(months,sales,marker="*",markersize=10,label="Sales Data ",lw=1,ls="--")
plt.title("Data Viz ")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True,alpha=0.2)
plt.show()
```

```python
import pandas as pd

df2 = {
    "WorkoutHours":[1,2,3,4,5,2.5,3.5,4.5],
    "CaloriesBurned":[120,180,250,320,410,210,290,360]
}

df2 = pd.DataFrame(df2)

df2
```

```python
plt.figure(figsize=(7,4),facecolor="white")

plt.scatter(df2["WorkoutHours"],df2["CaloriesBurned"],s=50,c="red",marker='^',lw=2,alpha=0.7)


plt.title("Scatter Plot ")
plt.grid(True,alpha=0.8)
plt.show()
```

```python
plt.figure(figsize=(7,4),facecolor="white")
plt.plot(df2["WorkoutHours"],df2["CaloriesBurned"],marker="*",markersize=10,label="Sales Data ",lw=1,ls="--")
plt.title("Data Viz ")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True,alpha=0.2)
plt.show()
```
